## SERIES

* It is a pandas data structure
* It is a 1D data structure
* In this data structure; each data have their own index (default/labelled)
* All values of series are of same type therefore series is homogeneous data structure

### Identification of series

* index 1 _____value 1
* index 2 _____value 2
* index 3 _____value 3
* index 4 _____value 4
* dtype : _____datatype of value

### Series ()

* It is a pandas module function which creates a series as per the defined data structure 
* identifier = pd.Series(data_structure , index = [List of Indexes])
* index is optional --> we'll use it only when we want labelled index

```python
pd.Series()
```

```python
a = pd.Series([1, 2, 3, 4, 5], index = ['a', 'b', 'c', 'd', 'e'])
a
```

```python
b = pd.Series([6,7,8,9,10]) # SINCE WE DIDN'T PROVIDE INDEX, IT WILL TAKE DEFAULT INDEX 0,1,2,3,4
b
```

```python
a+10
```
