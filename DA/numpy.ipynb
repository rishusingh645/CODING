#### identity()

1. It generates a square matrix (a 2D array where number of rows and columns are same), with "1" on the main diagonal and zeros elsewhere
2. Syntax --> np.identity(n) --> n is number of rows(and columns)
3. dtype is np.float64 (default)

```python
print(np.identity(5))
```

#### eye()

1. This is a more flexible version of identity()
2. While it also puts ones on the main diagonal & zeros elsewhere.
3. It allows us to create non square matrices & lets us shift the diagonal of "1" up or down
4. syntax --> np.eye(M, M = None , K = 0) 
    * N = number of rows
    * M = number of columns (If None, defaults to N, making the matrix square)
    * K = index of diagonal
        * 0 is the main diagonal
        * positive value shifts it up
        * negative value shifts it down

```python
print(np.eye(5))
```

```python
print(np.eye(5,4))
```

```python
print(np.eye(5,None,2))
```

```python
print(np.eye(5,None,-1))
```

### STACKING ARRAYS

```python
import numpy as np
a = np.random.randint(1,100,size = (2,3))
b = np.random.randint(1,100,size = (2,3))
print(a)
```

```python
print(b)
```

* There are two types of stacking.
    * HORIZONTAL STACK
        * hstack
        * arrays will join one after other (column stacking)
        * for horizontal stacking number of rows should remain same
    * VERTICAL STACK 
        * vstack 
        * arrays will join below one other 
        * for vertical stacking columns should remain same

```python
print(a)
print()
print(b)
print()
print(np.hstack((a,b))) # ROWS ARE SAME
```

```python
a = np.random.randint(1,100,size = (4,3))
b = np.random.randint(1,100,size = (2,3))
print(a)
print()
print(b)
```

```python
np.hstack((a,b)) # ROWS ARE NOT SAME, SO IT WILL THROW AN ERROR
```

```python
np.vstack((a,b)) # COLUMNS ARE SAME, SO IT WILL WORK FINE
```

```python
a = np.random.randint(1,100,size = (2,3))
b = np.random.randint(1,100,size = (2,4))
print(a)
print()
print(b)
print()
print(np.vstack((a,b))) # COLUMNS ARE NOT SAME, SO IT WILL THROW AN ERROR
```

* We can also perform the same operation using np.stack()
* SYNTAX:
        np.stack((a,b),axis = (0/1))

### where()

* It is used for data transformation
* It is used for data encoding ()

```python
marks = np.random.randint(1,100,5,dtype = np.int64)
print(marks)
```

```python
np.where(marks>50,"Pass","Fail")
```

```python
result = np.where(marks > 50,"Pass","Fail")
print(result)
```

### SELECT()

* It is used when we want to assign data by considering variouse conditions.
* IF-ELIF-ELSE wala hisaab

```python
grade = np.select(
    [marks >= 90, marks >= 75, marks >= 50, marks >= 33, marks < 33],
    ["A", "B", "C", "D", "F"],
    default="F"
)
grade
```
